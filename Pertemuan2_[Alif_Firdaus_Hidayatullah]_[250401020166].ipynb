Q1: Berapa jumlah baris dan kolom dataset Titanic?

```python
print(df.shape, df.columns.tolist())
```

Q2: Kolom mana yang memiliki nilai kosong
(missing)?

```python
missing = df.isnull().sum()
percent = (missing / len(df)) * 100

pd.DataFrame({
    'Missing': missing,
    'Percent (%)': percent
})
```

Q3: Berapa persen penumpang yang selamat
(Survived=1)?

```python

df['Survived'].mean() * 100
```

Q4: Rata-rata usia penumpang wanita kelas 1 yang
selamat?

```python
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
wanita_kelas1.shape[0], wanita_kelas1['Age'].mean()
```
